```python
import os

os.chdir('/content/movie-recommendation-system')

print("Project directory:", os.getcwd())
print("Dataset files:", os.listdir('dataset'))
print("Model files:", os.listdir('models'))
```

```python
# 1. Import Libraries

import pandas as pd
import pickle

from pathlib import Path
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# 2. Set Project Paths

ROOT_DIR = Path.cwd()

DATASET_DIR = ROOT_DIR / 'dataset'
MODELS_DIR = ROOT_DIR / 'models'
```

```python
# 3. Load Processed Dataset

df = pd.read_csv(DATASET_DIR / 'processed_movies.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# 4. Load Model Artifacts

with open(MODELS_DIR / 'tfidf_matrix.pkl', 'rb') as file:
    tfidf_matrix = pickle.load(file)

with open(MODELS_DIR / 'indices.pkl', 'rb') as file:
    indices = pickle.load(file)

with open(MODELS_DIR / 'tfidf.pkl', 'rb') as file:
    tfidf = pickle.load(file)

print("Model artifacts loaded successfully!")
```

```python
# 5. Calculate Cosine Similarity and Recommend Movies

def recommend(title, n=10):

    if title not in indices:
        return ['Movie not found']

    idx = indices[title]

    sim_score = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    similar_idx = sim_score.argsort()[::-1][1:n+1]

    return df['title'].iloc[similar_idx]
```

```python
# 6. Test Recommendations

recommend('Toy Story')
```

```python
# 7. Test Another Movie

recommend('Avatar')
```

```python
# 8. Test Invalid Movie Title

recommend('This Movie Does Not Exist')
```
